```python
# Import Required Libraries
import os
import subprocess
import sys
from pathlib import Path

# Define the Script Path and Runtime
workspace = Path.cwd()
script_path = workspace / 'script.sl'
print(f'Workspace: {workspace}')
print(f'Script path: {script_path}')

# If a compatible runtime exists, use it. Otherwise, report that the interpreter is unavailable.
runtime_candidates = ['python', 'py', 'powershell', 'pwsh']
runtime = None
for candidate in runtime_candidates:
    try:
        subprocess.run([candidate, '--version'], check=True, stdout=subprocess.PIPE, stderr=subprocess.PIPE)
        runtime = candidate
        break
    except Exception:
        continue

print(f'Runtime detected: {runtime or "none"}')
```

```python
# Execute the .sl Script
if not script_path.exists():
    raise FileNotFoundError(f'Missing script file: {script_path}')

if runtime == 'powershell' or runtime == 'pwsh':
    command = [runtime, '-ExecutionPolicy', 'Bypass', '-File', str(workspace / 'interpreter.ps1'), str(script_path)]
else:
    command = [runtime or 'python', str(workspace / 'interpreter.py'), str(script_path)]

result = subprocess.run(command, cwd=workspace, capture_output=True, text=True)
print('Command:', ' '.join(command))
```

```python
# Capture and Display Output
print('STDOUT:')
print(result.stdout)
print('STDERR:')
print(result.stderr)
print(f'Return code: {result.returncode}')
```

```python
# Handle Errors and Exit Codes
if result.returncode != 0:
    print('The script did not finish successfully.')
else:
    print('The script finished successfully.')
```
